# removing values 

```python
# Numbers of rows before removing null values
print("Rows before removing null values:",df.shape[0])

# Remove rows containing null values
df = df.dropna()

# Number of rows after removing null values
print("Rows after removing null values:",df.shape[0])
```

```python
print("Number of duplicate rows:",df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:",df.shape)
```

```python
# data cleaning and preprocessing
df["Order date"] = pd.to_datetime(df["Order Date"])
df["Ship date"] = pd.to_datetime(df["Ship Date"])
df[["Order date","Ship date"]].dtypes
```

```python
text_columns = ["Category","Sub-Category","Segment"]

for column in text_columns:
    df[column]=(
        df[column]
        .astype(str)
        .str.strip()
        .str.title()
    )
```

```python
for column in text_columns:
    print("\n,column")
    print(df[column].unique())
```

```python
# Final data check
print("Final shape:",df.shape)

print("\nTotal Missing Values:",df.isnull().sum().sum())

print("Total Duplicate Rows:",df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

# Sales and profit Analysis


```python
# Category wise sale
category_sales = (
    df.groupby("Category")["Sales"]
   .sum()
   .sort_values(ascending=False)
)

category_sales
```

```python
# Category wise profit
category_profit = (
    df.groupby("Category")["Sales"]
   .sum()
   .sort_values(ascending=False)
)

category_profit
```

```python
# Sub category wise profit
sub_category_profit = (
    df.groupby("Category")["Sales"]
   .sum()
   .sort_values(ascending=False)
)

sub_category_profit
```

```python
# REGION WISE PROFIT
region_profit = (
    df.groupby("Region")["Profit"]
   .sum()
   .sort_values(ascending=False)
)

region_profit
```

```python
numerical_df = df.select_dtypes(include=np.number)

correlation_matrix = numerical_df.corr()

correlation_matrix
```
