```python
print("--- 1. Check Positive Number ---")
num1 = int(input("Enter a number: "))
if num1 > 0:
    print(num1, "is Positive")
else:
    print(num1, "is Negative or Zero")

print("\n--- 2. Check Even or Odd ---")
num2 = int(input("Enter another number: "))
if num2 % 2 == 0:
    print(num2, "is Even")
else:
    print(num2, "is Odd")

print("\n--- 3. Assign Grade ---")
marks = int(input("Enter student marks out of 100: "))
if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 50:
    grade = "C"
else:
    grade = "Fail"
print("Grade:", grade)

print("\n--- 4. Login Verification---")
username = input("Enter username: ")
password = input("Enter password: ")
if username == "admin":
    if password == "1234":
        print("Login Successful")
    else:
        print("Wrong Password")
else:
    print("Wrong Username")

print("\n--- 5. Voting Eligibility ---")
age = int(input("Enter your age: "))
print("Eligible to vote") if age >= 18 else print("Not Eligible to vote")

print("\n--- 6. For Loop: 1 to 5 ---")
for i in range(1, 6):
    print(i, end=" ")

print("\n\n--- 7. While Loop: 1 to 5 ---")
j = 1
while j <= 5:
    print(j, end=" ")
    j += 1

print("\n\n--- 8. Break Statement ---")
for k in range(1, 10):
    if k == 5:
        break
    print(k, end=" ")

print("\n\n--- 9. Continue Statement ---")
for k in range(1, 10):
    if k == 5:
        continue
    print(k, end=" ")

print("\n\n--- 10. Pass Statement ---")
for k in range(1, 6):
    if k == 3:
        pass
    else:
        print(k, end=" ")

print("\n\n--- 11. Match-Case for Day ---")
day = int(input("Enter day number 1-7: "))
match day:
    case 1: print("Monday")
    case 2: print("Tuesday")
    case 3: print("Wednesday")
    case 4: print("Thursday")
    case 5: print("Friday")
    case 6: print("Saturday")
    case 7: print("Sunday")
    case _: print("Invalid day number")

print("\n--- Program End ---")
```

```python
def greet():
    print("Hello! Welcome to Python Functions.")
greet()

# 2. Function with arguments
def add(a, b):
    print("Sum =", a + b)
add(10, 20)

# 3. Function without arguments but with return value
def get_welcome():
    return "Welcome to learn python."
a = get_welcome()
print(a)

# 4. Function with arguments and return value
def multiply(a, b):
    return a * b
print("Product =", multiply(5, 4))

# 5. Function with default argument
def country(name="India"):
    print("Country:", name)
country()
country("Japan")

# 6. Function with keyword arguments
def student(name, age):
    print("Name:", name)
    print("Age:", age)
student(age=18, name="Ravi")

# 7. Function with variable-length arguments
def total_marks(*marks):
    total = sum(marks)
    print("Total Marks =", total)
total_marks(80, 90, 85, 95)

# 8. Lambda function
square = lambda x: x * x
print("Square =", square(6))


# 9. Recursive function for factorial
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)
print("Factorial =", factorial(5))


# 10. Nested function
def outer():
    print("Outer Function")
    
    def inner():
        print("Inner Function")
    
    inner()
outer()
```

```python
#Handling Missing Data using Pandas

import pandas as pd
import numpy as np

# Create a DataFrame containing missing values
data = {
    'Name': ['Arun', 'Bala', 'Chitra', 'Divya', 'Elango'],
    'Marks': [85, np.nan, 78, np.nan, 92],
    'Age': [20, 21, np.nan, 22, 21],
    'City': ['Chennai', 'Madurai', np.nan, 'Salem', 'Trichy']
}
df = pd.DataFrame(data)

# Display the original DataFrame
print("Original DataFrame:")
print(df)

# Check missing values using isnull()
print("\nMissing values (True/False):")
print(df.isnull())

# Count missing values in each column
print("\nCount of missing values per column:")
print(df.isnull().sum())

# Fill all missing values with a default value using fillna()
df_filled_default = df.fillna("Unknown")
print("\nAfter fillna() with default value 'Unknown':")
print(df_filled_default)

# Fill missing values in Marks column with column mean using fillna()
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
print("\nAfter fillna() - Marks column filled with mean:")
print(df)

# Forward fill - CORRECTED
df_ffill = df.ffill()
print("\nAfter ffill() - Forward Fill:")
print(df_ffill)

# Backward fill - CORRECTED
df_bfill = df.bfill()
print("\nAfter bfill() - Backward Fill:")
print(df_bfill)

# Remove rows with missing values using dropna()
df_dropped = df.dropna()
print("\nAfter dropna() - rows with missing values removed:")
print(df_dropped)

# Replace a specified value using replace()
df_replaced = df.replace("Chennai", "Coimbatore")
print("\nAfter replace() - 'Chennai' replaced with 'Coimbatore':")
print(df_replaced)

# Display DataFrame info
print("\nDataFrame Info:")
df.info()
```

```python
# Grouping and Aggregation Data using pandas

import pandas as pd

# Creating DataFrame
data = {
    "Department": ["HR", "IT", "HR", "IT", "Sales", "Sales"],
    "Employee": ["Amit", "Rahul", "Priya", "Ankit", "Sakshi", "Miya"],
    "Salary": [30000, 50000, 35000, 60000, 45000, 40000],
    "Experience": [2, 5, 3, 6, 4, 2]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Grouping by Department
grouped = df.groupby("Department")

# Sum Aggregation
print("\nSum of Salary by Department:")
print(grouped["Salary"].sum())

# Mean Aggregation
print("\nAverage Salary by Department:")
print(grouped["Salary"].mean())

# Maximum Salary
print("\nMaximum Salary by Department:")
print(grouped["Salary"].max())

# Minimum Salary
print("\nMinimum Salary by Department:")
print(grouped["Salary"].min())

# Multiple Aggregations
print("\nMultiple Aggregations:")
print(
    grouped["Salary"].agg(
        ["sum", "mean", "max", "min"]
    )
)

# Grouping Multiple Columns
print("\nGrouping by Department and Experience:")

group_multi = df.groupby(
    ["Department", "Experience"]
)["Salary"].sum()

print(group_multi)

# Iterating Groups
print("\nDisplaying Each Group:")

for name, group in grouped:
    print("\nDepartment:", name)
    print(group)

# Sorting Values
print("\nSorted DataFrame by Salary:")

sorted_df = df.sort_values(by="Salary")
print(sorted_df)
```

```python
# PRACTICAL 18
# Basic Data Cleaning and Preprocessing

import pandas as pd
import numpy as np

# Loading CSV File
file_path = r"students.csv"


# 1. Load the CSV dataset
#df = pd.read_csv(file_path)
df = pd.read_csv(file_path, sep=",", quoting=csv.QUOTE_NONE)

# 2. Display original dataset
print("========== ORIGINAL DATASET ==========")
print(df)

# 3. Display dataset information
print("\n========== ORIGINAL INFORMATION ==========")
df.info()

# 4. Check missing values
print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())

# 5. Fill missing Age values with mean
if "Age" in df.columns:
    df["Age"] = df["Age"].fillna(df["Age"].mean())

# 6. Fill missing Marks values with mean
if "Marks" in df.columns:
    df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# 7. Fill missing City values with "Unknown"
if "City" in df.columns:
    df["City"] = df["City"].fillna("Unknown")

# 8. Remove duplicate records
df = df.drop_duplicates()

print("\n========== AFTER REMOVING DUPLICATES ==========")
print(df)

# 9. Remove extra spaces from Name column
if "Name" in df.columns:
    df["Name"] = df["Name"].str.strip()

# 10. Convert Name column to lowercase
if "Name" in df.columns:
    df["Name"] = df["Name"].str.lower()

# 11. Convert City column to lowercase
if "City" in df.columns:
    df["City"] = df["City"].str.lower()

# 12. Convert Age into integer
if "Age" in df.columns:
    df["Age"] = df["Age"].astype(int)

# 13. Rename Marks column as Score
if "Marks" in df.columns:
    df.rename(columns={"Marks": "Score"}, inplace=True)

# 14. Sort dataset by Score
if "Score" in df.columns:
    df = df.sort_values(by="Score", ascending=False)

# 15. Reset the index
df = df.reset_index(drop=True)

# 16. Display cleaned dataset
print("\n========== CLEANED DATASET ==========")
print(df)

# 17. Check missing values after cleaning
print("\n========== MISSING VALUES AFTER CLEANING ==========")
print(df.isnull().sum())

# 18. Display final dataset information
print("\n========== FINAL DATASET INFORMATION ==========")
df.info()
```

```python
# PRACTICAL 19
# Calculate Basic Statistics from a Dataset

import pandas as pd
import numpy as np

# Loading CSV File
file_path = r"students.csv"


# 1. Load the CSV dataset
#df = pd.read_csv(file_path)
df = pd.read_csv(file_path, sep=",", quoting=csv.QUOTE_NONE)

# 2. Display the dataset
print("========== DATASET ==========")
print(df)

# 3. Select Marks column
marks = df["Marks"]

# 4. Remove missing values
marks = marks.dropna()

# 5. Count number of observations
count = marks.count()

# 6. Calculate total marks
total = marks.sum()

# 7. Calculate Mean
mean = marks.mean()

# 8. Calculate Median
median = marks.median()

# 9. Calculate Mode
mode = marks.mode()

# 10. Calculate Standard Deviation
std = marks.std()

# 11. Calculate Variance
variance = marks.var()

# 12. Find Minimum
minimum = marks.min()

# 13. Find Maximum
maximum = marks.max()

# 14. Calculate Range
data_range = maximum - minimum

# 15. Calculate First Quartile
q1 = marks.quantile(0.25)

# 16. Calculate Third Quartile
q3 = marks.quantile(0.75)

# 17. Display results
print("\n========== BASIC STATISTICS ==========")

print("Number of Observations :", count)
print("Total Marks            :", total)
print("Mean                   :", mean)
print("Median                 :", median)
print("Mode                   :", mode.tolist())
print("Standard Deviation     :", std)
print("Variance               :", variance)
print("Minimum                :", minimum)
print("Maximum                :", maximum)
print("Range                  :", data_range)
print("First Quartile (Q1)    :", q1)
print("Third Quartile (Q3)    :", q3)
```
