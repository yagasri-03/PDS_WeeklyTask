```python
# Fill numerical columns with mean
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
print(df.dtypes)
print(df.isnull().sum())
```

```python
# Convert date into datetime
df["date"] = pd.to_datetime(
    df["date"],
    errors="coerce",
    utc=True,
    format="mixed"
)

print(df["date"].dtype)
print(df["date"].head())
```

```python
# Check missing values
print("Missing values:")
print(df.isnull().sum())

# Remove rows with missing essential stock information
df = df.dropna(
    subset=["date", "open", "high", "low", "close"]
)

# Fill missing volume with median
df["volume"] = df["volume"].fillna(
    df["volume"].median()
)

# Remove duplicates
df = df.drop_duplicates()

# Sort data by date
df = df.sort_values("date").reset_index(drop=True)

print("After cleaning:")
print(df.isnull().sum())
print(df.shape)
```

```python
df["daily_price_change"] = (
    df["close"] - df["open"]
)

print(df[[
    "date",
    "open",
    "close",
    "daily_price_change"
]].head())
```

```python
df["daily_return"] = (
    df["close"].pct_change() * 100
)

print(df[[
    "date",
    "close",
    "daily_return"
]].head())
```

```python
df["price_range"] = df["high"] - df["low"]

print(df[[
    "date",
    "high",
    "low",
    "price_range"
]].head())
```

```python
df["SMA_20"] = (
    df["close"].rolling(window=20).mean()
)

df["SMA_50"] = (
    df["close"].rolling(window=50).mean()
)

print(df[[
    "date",
    "close",
    "SMA_20",
    "SMA_50"
]].tail())
```

```python
print("Mean Return:",
      df["daily_return"].mean())

print("Median Return:",
      df["daily_return"].median())

print("Variance:",
      df["daily_return"].var())

print("Standard Deviation:",
      df["daily_return"].std())

print("Average Volume:",
      df["volume"].mean())
```

```python
df["rolling_volatility"] = (
    df["daily_return"].rolling(window=20).std()
)

print(df[[
    "date",
    "daily_return",
    "rolling_volatility"
]].tail())
```

```python
for col in ["daily_return", "volume"]:

    Q1 = df[col].quantile(0.25)

    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR

    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]

    print("\nColumn:", col)

    print("Lower Bound:", lower)

    print("Upper Bound:", upper)

    print("Number of Outliers:", len(outliers))
```

```python
# Select last 60 trading days
plot_df = df.tail(60).reset_index(drop=True)

plt.figure(figsize=(14, 6))

for i, row in plot_df.iterrows():

    color = "green" if row["close"] >= row["open"] else "red"

    # High and Low
    plt.vlines(
        i,
        row["low"],
        row["high"],
        color=color
    )

    # Open price
    plt.hlines(
        row["open"],
        i - 0.3,
        i,
        color=color
    )

    # Close price
    plt.hlines(
        row["close"],
        i,
        i + 0.3,
        color=color
    )

plt.title("Shopify Stock OHLC Chart")
plt.xlabel("Trading Days")
plt.ylabel("Stock Price")
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df["date"],
    df["close"],
    label="Closing Price"
)

plt.plot(
    df["date"],
    df["SMA_20"],
    label="20-Day Moving Average"
)

plt.plot(
    df["date"],
    df["SMA_50"],
    label="50-Day Moving Average"
)

plt.title("Shopify Stock Moving Average Analysis")

plt.xlabel("Date")
plt.ylabel("Stock Price")

plt.legend()
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    df["date"],
    df["volume"]
)

plt.title("Shopify Trading Volume Trends")

plt.xlabel("Date")
plt.ylabel("Trading Volume")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    df["daily_return"].dropna(),
    bins=30,
    kde=True
)

plt.title("Shopify Daily Returns Distribution")

plt.xlabel("Daily Return (%)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.kdeplot(
    df["daily_return"].dropna(),
    fill=True
)

plt.title("Shopify Daily Returns KDE Plot")

plt.xlabel("Daily Return (%)")

plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    df["date"],
    df["rolling_volatility"]
)

plt.title("Shopify 20-Day Rolling Volatility")

plt.xlabel("Date")
plt.ylabel("Volatility (%)")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 7))

corr = df.select_dtypes(
    include=np.number
).corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Shopify Stock Correlation Heatmap")

plt.show()
```
