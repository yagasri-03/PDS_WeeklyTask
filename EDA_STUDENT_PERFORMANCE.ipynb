```python
# Correlation Matrix
corr = df[["math score", "reading score", "writing score"]].corr()
print(corr)
```

```python
# Grouped Bar Chart - Parental Education
df1 = df.groupby("parental level of education")[["math score", "reading score", "writing score"]].mean()
df1.plot(kind="bar")
plt.show()
```

```python
# Box Plot - Test Preparation
sns.boxplot(x="test preparation course", y="reading score", data=df)
plt.show()

# Test Preparation - Writing
sns.boxplot(data=df, x="test preparation course", y="writing score")
plt.show()
```

```python
# Scatter Plot - Reading and Writing
sns.scatterplot(x="reading score", y="writing score", data=df)
plt.show()
```

```python
# Correlation Heatmap
sns.heatmap(corr, annot=True)
plt.show()
```

```python
# Lunch Type
sns.barplot(data=df, x="lunch", y="math score")
plt.show()
```
