```python
# 17. Create a New Feature - Length of Stay

df["Length of Stay"] = (
    pd.to_datetime(df["Discharge Date"]) - 
    pd.to_datetime(df["Date of Admission"])
).dt.days

df[["Date of Admission", "Discharge Date", "Length of Stay"]].head()
```

```python
# 18. Create a New Feature - Admission Year

df["Admission Year"] = pd.to_datetime(
    df["Date of Admission"]
).dt.year

df[["Date of Admission", "Admission Year"]].head()
```

```python
# 19. Create a New Feature - Admission Month

df["Admission Month"] = pd.to_datetime(
    df["Date of Admission"]
).dt.month

df[["Date of Admission", "Admission Month"]].head()
```

```python
# 20. Average Billing Amount by Admission Year

df.groupby("Admission Year")["Billing Amount"].mean()
```

```python
# 28. Prepare Monthly Admission Data

monthly_admissions = (
    df.groupby("Admission Month")["Name"]
    .count()
    .reset_index()
)

monthly_admissions
```

```python
plt.figure(figsize=(9,5))

sns.lineplot(
    data = monthly_admissions,
    x="Admission Month",
    y="Name",
    marker="o"
)

plt.title("Monthly Number of Admissions")
plt.xlabel("Admission Month")
plt.ylabel("Number of Admissions")

plt.show()
```

```python
# 1. BAR CHART
# Patients by Medical Condition
# ==========================================

condition_count = df["Medical Condition"].value_counts()

plt.figure(figsize=(8,5))
plt.bar(condition_count.index, condition_count.values)

plt.title("Patients by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)

plt.show()
```

```python
# 2. LINE CHART
df.groupby("Medical Condition")["Billing Amount"].mean().plot(kind="line", marker="o")

plt.title("Average Billing Amount")
plt.xlabel("Medical Condition")
plt.ylabel("Billing Amount")
plt.show()
```

```python
# 2. LINE CHART
df.groupby("Medical Condition")["Billing Amount"].mean().plot(kind="line", marker="o")

plt.title("Average Billing Amount")
plt.xlabel("Medical Condition")
plt.ylabel("Billing Amount")
plt.show()
```

```python
# 3. PIE CHART
df["Gender"].value_counts().plot(kind="pie", autopct="%1.1f%%")

plt.title("Gender Distribution")
plt.ylabel("")
plt.show()
```

```python
# 4. Scatter Plot

plt.scatter(df["Age"], df["Room Number"])

plt.title("Age vs Room Number")
plt.xlabel("Age")
plt.ylabel("Room Number")

plt.show()
```

```python
# 5. BOX PLOT
sns.boxplot(x="Admission Type", y="Billing Amount", data=df)

plt.title("Billing Amount by Admission Type")
plt.show()
```
